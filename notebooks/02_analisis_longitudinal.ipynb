# Análisis Longitudinal de Investigadores

Este análisis estudia la evolución temporal de los investigadores
participantes en convocatorias de MinCiencias.

```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
from pathlib import Path

base_dir = Path.cwd().resolve().parent
ruta_csv = base_dir / "datos" / "tarea_join" / "investigadores_consolidado.csv"
ruta_xlsx = base_dir / "datos" / "raw" / "investigadores_consolidado.xlsx"

if ruta_csv.exists():
    df = pd.read_csv(ruta_csv, low_memory=False)
elif ruta_xlsx.exists():
    df = pd.read_excel(ruta_xlsx)
else:
    raise FileNotFoundError(
        "No se encontró el archivo de investigadores en datos/tarea_join ni en datos/raw"
    )

df.head()
```

```python
# Análisis por convocatoria
if "ANO_CONVO" not in df.columns:
    raise KeyError("La columna ANO_CONVO no existe en el dataset")

serie_ano = pd.to_numeric(df["ANO_CONVO"], errors="coerce").dropna().astype(int)
convocatorias = serie_ano.value_counts().sort_index()

ax = convocatorias.plot(marker="o", figsize=(8, 4), grid=True)
ax.set_title("Investigadores por convocatoria")
ax.set_xlabel("Año")
ax.set_ylabel("Cantidad")
plt.tight_layout()
plt.show()

convocatorias
```
