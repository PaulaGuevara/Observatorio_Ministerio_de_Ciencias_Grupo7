# Análisis Longitudinal de Investigadores

Este análisis estudia la evolución temporal de los investigadores
participantes en convocatorias de MinCiencias.

```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
ruta = "../datos/raw/investigadores_consolidado.xlsx"

df = pd.read_excel(ruta)
```

```python
#Análisis por convocatoria:
convocatorias = df["ANO_CONVO"].value_counts().sort_index()

convocatorias.plot(marker="o")

plt.title("Investigadores por convocatoria")
plt.xlabel("Año")
plt.ylabel("Cantidad")
plt.show()
```
